```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from pathlib import Path

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    precision_recall_curve,
    average_precision_score
)

print("All libraries imported successfully!")
```

```python
BASE_DIR = Path.cwd().parent

DATA_DIR = BASE_DIR / "datasets"
MODEL_DIR = BASE_DIR / "models"
OUTPUT_DIR = BASE_DIR / "outputs"

MODEL_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)

VIDEO_PATH = DATA_DIR / "video_telemetry.csv"
SYSTEM_PATH = DATA_DIR / "system_events.csv"

print("Project directory:", BASE_DIR)
print("Video file exists:", VIDEO_PATH.exists())
print("System file exists:", SYSTEM_PATH.exists())
```

```python
video_df = pd.read_csv(VIDEO_PATH)

system_df = pd.read_csv(SYSTEM_PATH)

video_df["timestamp"] = pd.to_datetime(video_df["timestamp"])

system_df["timestamp"] = pd.to_datetime(system_df["timestamp"])

video_df = video_df.sort_values("timestamp").reset_index(drop=True)

system_df = system_df.sort_values("timestamp").reset_index(drop=True)

print("VIDEO DATASET")
print(video_df.shape)

display(video_df.head())

print("SYSTEM EVENTS DATASET")
print(system_df.shape)

display(system_df.head())
```

```python
print("VIDEO DATA INFORMATION")

video_df.info()

print("\nSYSTEM EVENTS INFORMATION")

system_df.info()

print("\nMISSING VALUES")

print(video_df.isnull().sum())

print("\nSYSTEM EVENT MISSING VALUES")

print(system_df.isnull().sum())

print("\nCHEATING LABEL DISTRIBUTION")

print(system_df["is_cheating"].value_counts())
```

```python
# Handle missing sensor values
sensor_columns = ["eye_gaze_angle", "audio_db"]

for column in sensor_columns:

    video_df[column] = video_df[column].interpolate(
        method="linear",
        limit=10,
        limit_direction="both"
    )

# Fill any remaining sensor gaps with column median
for column in sensor_columns:

    video_df[column] = video_df[column].fillna(
        video_df[column].median()
    )

print("Missing sensor values after preprocessing:")

print(video_df[sensor_columns].isnull().sum())

print("\nPreprocessing completed!")
```

```python
# 10-second rolling average
video_df["gaze_rolling_10s"] = (
    video_df["eye_gaze_angle"]
    .rolling(window=10, min_periods=1)
    .mean()
)

video_df["audio_rolling_10s"] = (
    video_df["audio_db"]
    .rolling(window=10, min_periods=1)
    .mean()
)

# Gaze variation
video_df["gaze_std_10s"] = (
    video_df["eye_gaze_angle"]
    .rolling(window=10, min_periods=1)
    .std()
    .fillna(0)
)

display(video_df.head(15))

print("Feature engineering completed!")
```

```python
# Keep event labels at their original timestamps
events = system_df.copy()

# Match each event to the latest available video observation
merged_df = pd.merge_asof(
    events.sort_values("timestamp"),
    video_df.sort_values("timestamp"),
    on="timestamp",
    direction="backward",
    tolerance=pd.Timedelta("2s")
)

print("Merged dataset shape:", merged_df.shape)

print("\nMissing values after merging:")

print(merged_df.isnull().sum())

display(merged_df.head())
```

```python
features = [
    "tab_switches",
    "eye_gaze_angle",
    "audio_db",
    "gaze_rolling_10s",
    "audio_rolling_10s",
    "gaze_std_10s"
]

# Keep only rows with available features and labels
model_df = merged_df.dropna(
    subset=features + ["is_cheating"]
).copy()

X = model_df[features]

y = model_df["is_cheating"].astype(int)

print("Final dataset shape:", X.shape)

print("\nClass distribution:")

print(y.value_counts())
```

```python
# Chronological train-test split
split_index = int(len(model_df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())

# Train Random Forest
model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    min_samples_leaf=2
)

model.fit(X_train, y_train)

print("\nModel training completed successfully!")
```

```python
# Predict probabilities
y_probability = model.predict_proba(X_test)[:, 1]

# Custom threshold
threshold = 0.90

y_pred = (y_probability >= threshold).astype(int)

print("Decision threshold:", threshold)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)

print("Accuracy:", accuracy_score(y_test, y_pred))

print(
    "Precision:",
    precision_score(y_test, y_pred, zero_division=0)
)

print(
    "Recall:",
    recall_score(y_test, y_pred, zero_division=0)
)

print(
    "F1 Score:",
    f1_score(y_test, y_pred, zero_division=0)
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Innocent", "Suspicious"]
)

disp.plot(cmap="Blues")

plt.title("Project Panopticon - Confusion Matrix")

plt.savefig(
    OUTPUT_DIR / "confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_probability
)

ap = average_precision_score(
    y_test,
    y_probability
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall,
    precision,
    label=f"Average Precision = {ap:.3f}"
)

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title("Precision-Recall Curve")

plt.legend()
plt.grid(True)

plt.savefig(
    OUTPUT_DIR / "precision_recall_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
joblib.dump(
    model,
    MODEL_DIR / "panopticon_model.pkl"
)

joblib.dump(
    features,
    MODEL_DIR / "feature_columns.pkl"
)

print("Model saved successfully!")
print("Feature columns saved successfully!")
```
